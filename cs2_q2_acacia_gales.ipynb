```python
# Movie Rating Analyzer
movies = ["Inception", "The Matrix", "Interstellar", "Titanic", "Coco"]
ratings = [8.8, 8.7, 8.6, 7.9, 8.4]

def show_ratings():
    print("\n--- Movie Ratings ---")
    for i in range(len(movies)):
        print(f"{i + 1}. {movies[i]} - {ratings[i]}")

def search_rating():
    name = input("Enter a movie name: ")
    for i in range(len(movies)):
        if movies[i].lower() == name.lower():
            print(movies[i], "has a rating of", ratings[i])
            return
    print("Movie not found.")

def average_rating():
    average = sum(ratings) / len(ratings)
    print("Average rating:", round(average, 2))

def update_rating():
    show_ratings()
    number = int(input("Enter the number of the movie to update: "))
    new_rating = float(input("Enter the new rating: "))
    ratings[number - 1] = new_rating   # number - 1 because lists start at 0
    print("Rating updated!")

while True:
    print("\n=== Movie Rating Analyzer ===")
    print("1. Display Ratings")
    print("2. Search Rating")
    print("3. Compute Average Rating")
    print("4. Update Rating")
    print("5. Display Updated Ratings")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        show_ratings()
    elif choice == "2":
        search_rating()
    elif choice == "3":
        average_rating()
    elif choice == "4":
        update_rating()
    elif choice == "5":
        show_ratings()
    elif choice == "6":
        print("Goodbye!")
        break
    else:
        print("Please enter a number from 1 to 6.")
```
